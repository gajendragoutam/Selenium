```python
from selenium import webdriver
from selenium.webdriver.common.by import By
import  pandas as pd
import time
import warnings
warnings.filterwarnings("ignore")
```

```python
driver= webdriver.Chrome()
driver.get("https://www.google.co.in/maps/@21.7657658,79.7970339,14z?entry=ttu&g_ep=EgoyMDI0MTAxNi4wIKXMDSoASAFQAw%3D%3D")
driver
```

```python
put_location = driver.find_element(By.ID,"searchboxinput")
put_location.send_keys("Restaurant bhopal madhya pradesh")
```

```python
button = driver.find_element(By.CLASS_NAME,("google-symbols"))
button.click()
time.sleep(5)
```

```python
Rastaurant      =[]
Location_in     =[]
Ratings         =[]
Close_time      =[]
```

```python
Name = driver.find_elements(By.XPATH,'//div[@class="qBF1Pd fontHeadlineSmall "]')
for i in Name:
    Rastaurant.append(i.text)
```

```python
locations = driver.find_elements(By.XPATH,"//div[@class='W4Efsd']/span[3]")
for i in locations:
    Location_in.append(i.text)
time.sleep(3)
```

```python
ratings = driver.find_elements(By.XPATH,'//span[@class="MW4etd"]')
for i in ratings:
    text = i.text
    Ratings.append(text)
```

```python
close = driver.find_elements(By.XPATH,'//span[contains(text(), "Closes")]')
for i in close:
    clean_time = i.text.replace('\u202f', '').strip()
    Close_time.append(clean_time)
```

```python
len(Ratings), len(Rastaurant) , len(Location_in) , len(Close_time)
```

```python
Close_time += [None] * (40 - len(Close_time))
Location_in += [None] * (40 - len(Location_in))
```

```python
Dataframe = pd.DataFrame({
    "Rastaurant_Name": Rastaurant , 
    "Location": Location_in ,
    "Rating" : Ratings ,
    "Close_Time" : Close_time
    
})
```
